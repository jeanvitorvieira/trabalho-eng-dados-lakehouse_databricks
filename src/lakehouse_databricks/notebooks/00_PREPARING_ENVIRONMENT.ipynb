# 00 - Preparando o ambiente

Cria os schemas (databases) das camadas do Medalhão e o volume da Landing.

- `workspace.landing` + volume `workspace.landing.dados`: arquivos CSV brutos
- `workspace.bronze`: Delta sem transformação
- `workspace.silver`: Delta com Data Quality
- `workspace.gold`: modelo dimensional (Kimball)

O notebook é idempotente (`IF NOT EXISTS`), então pode rodar quantas vezes precisar.

```python
%sql

CREATE SCHEMA IF NOT EXISTS workspace.landing
COMMENT 'Schema/Database da landing zone (arquivos brutos em volume)';

CREATE VOLUME IF NOT EXISTS workspace.landing.dados
COMMENT 'Volume para os CSV brutos extraídos da fonte';

CREATE SCHEMA IF NOT EXISTS workspace.bronze
COMMENT 'Schema/Database para dados bronze (delta)';

CREATE SCHEMA IF NOT EXISTS workspace.silver
COMMENT 'Schema/Database para dados silver (delta)';

CREATE SCHEMA IF NOT EXISTS workspace.gold
COMMENT 'Schema/Database para dados gold (delta) - modelagem dimensional';
```
